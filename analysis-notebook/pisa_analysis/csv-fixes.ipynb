```python
import csv
```

```python
with open('../data/education-data/processed/melted_output.csv', newline='', encoding='utf-8') as infile, \
     open('../data/education-data/processed/melted_output_clean2.csv', 'w', newline='', encoding='utf-8') as outfile:
    reader = csv.reader(infile)
    writer = csv.writer(outfile)
    header = next(reader)
    writer.writerow(header)
    for row in reader:
        if len(row) >= 5 and not row[4].startswith('Unnamed'):
            writer.writerow(row)
```
