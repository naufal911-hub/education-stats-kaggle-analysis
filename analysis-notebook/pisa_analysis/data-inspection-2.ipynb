# Cleaning data

```python
import pandas as pd
```

```python
data_dir="../data/education-data/raw/edstats-csv-zip-32-mb-"
df_data=pd.read_csv(f"{data_dir}/EdStatsData.csv")
```

```python
id_vars=['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code']
year_cols=[col for col in df_data.columns if col not in id_vars ]
```

```python
df_long = df_data.melt(
    id_vars=id_vars,
    value_vars=year_cols,
    var_name='Year',
    value_name='Value'
)
```

```python
df_malaysia=df_long[df_long['Country Name']=='Malaysia']
```

```python
df_malaysia_clean = df_malaysia.dropna(subset=['Value']).copy()
```
