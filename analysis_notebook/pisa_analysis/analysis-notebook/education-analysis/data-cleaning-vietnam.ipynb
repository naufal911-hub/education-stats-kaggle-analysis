```python
import pandas as pd
```

```python
data_dir="../data/education-data/edstats-csv-zip-32-mb-"
df_data=pd.read_csv(f"{data_dir}/EdStatsData.csv")
```

```python
id_vars=['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code']
year_cols=[col for col in df_data.columns if col not in id_vars ]
```

```python
chunk_size = 20  # number of year-columns to melt at a time; tune down if it still dies
output_path = 'melted_output.csv'
```

```python
for i in range(0, len(year_cols), chunk_size):
    chunk_cols = year_cols[i:i + chunk_size]
    melted_chunk = df_data.melt(
        id_vars=id_vars,
        value_vars=chunk_cols,
        var_name='Year',
        value_name='Value'
    )
    
    melted_chunk.to_csv(
        output_path,
        mode='a',
        header=(i == 0),
        index=False
    )
    
    del melted_chunk
    print(f"Melted columns {i} to {i + chunk_size}")
```
